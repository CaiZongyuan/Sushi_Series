使用GLM voice 批量提取 speech token

```python
import sys
import os
sys.path.append('/root/LLMs/GLM-4-Voice')

import json
import torch
from dataclasses import dataclass
from typing import List, Dict
from tqdm import tqdm  # 导入 tqdm 库用于进度条

from speech_tokenizer.modeling_whisper import WhisperVQEncoder
from speech_tokenizer.utils import extract_speech_token
from transformers import WhisperFeatureExtractor

@dataclass
class TokenizerConfig:
    glm_tokenizer_path: str = "/root/LLMs/GLM-4-Voice/THUDM/glm-4-voice-tokenizer"

class GLMTokenizer:
    def __init__(self, config: TokenizerConfig):
        self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        self.config = config
        self._init_glm_voice()
    
    def _init_glm_voice(self):
        print("加载 GLM Voice 模型...")
        self.whisper_model = WhisperVQEncoder.from_pretrained(
            self.config.glm_tokenizer_path
        ).eval().to(self.device)
        self.glm_feature_extractor = WhisperFeatureExtractor.from_pretrained(
            self.config.glm_tokenizer_path
        )
        print("GLM Voice 模型加载完成。")
    
    def get_glm_tokens_batch(self, audio_paths: List[str]) -> List[List[int]]:
        """
        提取一个批次音频文件的语音 token
        
        Args:
            audio_paths (List[str]): 音频文件路径列表。
        
        Returns:
            List[List[int]]: 每个音频文件对应的 token 列表。
        """
        tokens = extract_speech_token(self.whisper_model, self.glm_feature_extractor, audio_paths)
        return tokens

def read_wav_scp(wav_scp_path: str) -> Dict[str, str]:
    """
    从 wav.scp 文件中读取音频文件路径
    
    Args:
        wav_scp_path (str): wav.scp 文件路径。
    
    Returns:
        Dict[str, str]: 音频文件名到音频文件路径的映射。
    """
    print(f"正在读取 wav.scp 文件: {wav_scp_path}...")
    wav_dict = {}
    with open(wav_scp_path, 'r') as f:
        for line in f:
            name, path = line.strip().split(maxsplit=1)
            wav_dict[name] = path
    print(f"成功读取 {len(wav_dict)} 条音频文件路径。")
    return wav_dict

def save_tokens_to_json(tokens_dict: Dict[str, List[int]], output_path: str):
    """
    将 token 保存到 JSON 文件中
    
    Args:
        tokens_dict (Dict[str, List[int]]): 音频文件名到 token 列表的映射。
        output_path (str): 输出 JSON 文件路径。
    """
    print(f"正在将结果保存到 JSON 文件: {output_path}...")
    with open(output_path, 'w') as f:
        json.dump(tokens_dict, f, indent=4)
    print(f"结果已成功保存到 {output_path}。")

def main(wav_scp_path: str, output_json_path: str):
    # 初始化 GLM Tokenizer
    print("初始化 GLM Tokenizer...")
    config = TokenizerConfig()
    glm_tokenizer = GLMTokenizer(config)
    
    # 读取 wav.scp 文件
    wav_dict = read_wav_scp(wav_scp_path)
    
    # 提取音频文件名和路径
    audio_names = list(wav_dict.keys())
    audio_paths = list(wav_dict.values())
    
    # 批量提取 speech token
    print("开始提取 speech token...")
    tokens_dict = {}
    for i in tqdm(range(0, len(audio_paths), batch_size), desc="处理进度"):
        batch_paths = audio_paths[i:i + batch_size]
        batch_names = audio_names[i:i + batch_size]
        batch_tokens = glm_tokenizer.get_glm_tokens_batch(batch_paths)
        for name, token in zip(batch_names, batch_tokens):
            tokens_dict[name] = token
    
    # 保存结果到 JSON 文件
    save_tokens_to_json(tokens_dict, output_json_path)

if __name__ == "__main__":
    wav_scp_path = "/root/LLMs/data/Emilia-Dataset/CosyTokens/wav.scp"  # 替换为你的 wav.scp 文件路径
    output_json_path = "/root/LLMs/data/Emilia-Dataset/CosyTokens/glm_output.json"  # 替换为你的输出 JSON 文件路径
    batch_size = 8  # 设置批量处理的大小
    print(f"批量处理大小设置为: {batch_size}")
    main(wav_scp_path, output_json_path)
```

将两个speech token文件合并成一个json文件

```python
import json
from typing import Dict, List
import torch
from tqdm import tqdm  # 导入 tqdm 库用于进度条

def load_cosyvoice_tokens(cosytoken_path: str) -> Dict[str, List[int]]:
    """
    从 .pt 文件中加载 cosyvoice_tokens
    
    Args:
        cosytoken_path (str): .pt 文件路径。
    
    Returns:
        Dict[str, List[int]]: 音频文件名到 cosyvoice_tokens 的映射。
    """
    print(f"正在加载 cosyvoice_tokens 文件: {cosytoken_path}...")
    cosyvoice_tokens = torch.load(cosytoken_path)
    print(f"成功加载 {len(cosyvoice_tokens)} 条 cosyvoice_tokens。")
    return cosyvoice_tokens

def load_glm_tokens(glm_token_path: str) -> Dict[str, List[int]]:
    """
    从 JSON 文件中加载 glm_tokens
    
    Args:
        glm_token_path (str): JSON 文件路径。
    
    Returns:
        Dict[str, List[int]]: 音频文件名到 glm_tokens 的映射。
    """
    print(f"正在加载 glm_tokens 文件: {glm_token_path}...")
    with open(glm_token_path, 'r') as f:
        glm_tokens = json.load(f)
    print(f"成功加载 {len(glm_tokens)} 条 glm_tokens。")
    return glm_tokens

def merge_tokens(glm_tokens: Dict[str, List[int]], cosyvoice_tokens: Dict[str, List[int]]) -> Dict[str, Dict[str, List[int]]]:
    """
    合并 glm_tokens 和 cosyvoice_tokens
    
    Args:
        glm_tokens (Dict[str, List[int]]): 音频文件名到 glm_tokens 的映射。
        cosyvoice_tokens (Dict[str, List[int]]): 音频文件名到 cosyvoice_tokens 的映射。
    
    Returns:
        Dict[str, Dict[str, List[int]]]: 合并后的数据，包含 name、glm_tokens 和 cosyvoice_tokens。
    """
    print("正在合并 glm_tokens 和 cosyvoice_tokens...")
    merged_tokens = {}
    for name in tqdm(glm_tokens, desc="合并进度"):
        if name in cosyvoice_tokens:
            merged_tokens[name] = {
                "name": name,
                "glm_tokens": glm_tokens[name],
                "cosyvoice_tokens": cosyvoice_tokens[name]
            }
        else:
            print(f"警告: {name} 在 cosyvoice_tokens 中未找到，跳过该文件。")
    print(f"成功合并 {len(merged_tokens)} 条数据。")
    return merged_tokens

def save_tokens_to_json(tokens_dict: Dict[str, Dict[str, List[int]]], output_path: str):
    """
    将合并后的 token 保存到 JSON 文件中
    
    Args:
        tokens_dict (Dict[str, Dict[str, List[int]]]): 合并后的数据。
        output_path (str): 输出 JSON 文件路径。
    """
    print(f"正在将结果保存到 JSON 文件: {output_path}...")
    with open(output_path, 'w') as f:
        json.dump(tokens_dict, f, indent=4)
    print(f"结果已成功保存到 {output_path}。")

def main(glm_token_path: str, cosytoken_path: str, output_json_path: str):
    # 加载 glm_tokens
    glm_tokens = load_glm_tokens(glm_token_path)
    
    # 加载 cosyvoice_tokens
    cosyvoice_tokens = load_cosyvoice_tokens(cosytoken_path)
    
    # 合并数据
    merged_tokens = merge_tokens(glm_tokens, cosyvoice_tokens)
    
    # 保存结果到 JSON 文件
    save_tokens_to_json(merged_tokens, output_json_path)

if __name__ == "__main__":
    glm_token_path = "/root/LLMs/data/Emilia-Dataset/CosyTokens/glm_output.json"  # 替换为你的 glm_tokens JSON 文件路径
    cosytoken_path = "/root/LLMs/data/Emilia-Dataset/CosyTokens/utt2speech_token.pt"  # 替换为你的 .pt 文件路径
    output_json_path = "/root/LLMs/data/Emilia-Dataset/CosyTokens/tokens.json"  # 替换为你的输出 JSON 文件路径
    main(glm_token_path, cosytoken_path, output_json_path)
```
