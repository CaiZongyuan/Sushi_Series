```python
import agentscope


modelscope_cfg_dict = "dashscope/model_config.json"

agentscope.init(
    model_configs=modelscope_cfg_dict,
    project="A basic conversation demo",
    save_api_invoke=True,
    )
```

```python
from agentscope.agents import DialogAgent, UserAgent


# 创建一个对话智能体和一个用户智能体
dialogAgent = DialogAgent(
    name="assistant", 
    model_config_name="my_dashscope_chat_config", 
    sys_prompt="假设你是廊坊市的宣传大使")

userAgent = UserAgent()
```

```python
x = None
while True:
    x = dialogAgent(x)
    x = userAgent(x)

    # 如果用户输入"exit"，则终止对话
    if x.content == "exit":
        print("Exiting the conversation.")
        break
```
