假设我们有从两个分布中的采样$X_0∼π_0$，$X_1∼π_1$(比如$X_0$是从$π_0$里出来的随机噪声， $X_1$是一个随机的数据(服从$π_1$))。我们把$X_0$和$X_1$用一个线性插值连接起来，得到

$X_t=tX_1+(1−t)X_0,  t∈[0,1]$

```python
import numpy as np
import matplotlib.pyplot as plt

# 假设X_0和X_1是二维空间中的点
X_0 = np.array([0, 0])
X_1 = np.array([1, 1])

# 创建t的数组，从0到1
t = np.linspace(0, 1, 100)

# 计算X_t
X_t = t[:, None] * X_1 + (1 - t[:, None]) * X_0

# 创建图形
plt.figure(figsize=(8, 6))

# 绘制X_t的轨迹
plt.plot(X_t[:, 0], X_t[:, 1], label='Trajectory of $X_t$',color='g')

# 绘制起点X_0和终点X_1
plt.plot(X_0[0], X_0[1], 'bo', markersize=10, label='$X_0$')
plt.plot(X_1[0], X_1[1], 'yo', markersize=10, label='$X_1$')

# 显示方向向量dX_t/dt = X_1 - X_0
dx = X_1 - X_0

plt.quiver(X_0[0], X_0[1], dx[0], dx[1], angles='xy', scale_units='xy', scale=9, color='r', label=r'$\frac{d}{dt}X_t = X_1 - X_0$')

# 设置图形属性
plt.xlabel('x')
plt.ylabel('y')

plt.title('Visualization of $X_t$ and its derivative')
plt.legend()
plt.grid(False)
plt.axis('equal')  # 等比例显示坐标轴，使得方向向量正确显示
plt.show()
```
